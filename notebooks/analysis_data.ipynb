```python
import pandas as pd

path = "../data/parks.json"
```

```python
df = pd.read_json(path)
df.info()
df.head()
```

```python
df["parks"].head(1)
```

```python
df_normalized = df.copy()

df_normalized = df_normalized.explode('parks').reset_index(drop=True)

df_normalized = pd.json_normalize(df_normalized['parks']).add_prefix(f"parks.")


df_final = (
    df_normalized
    .drop(columns=["parks"])
    .reset_index(drop=True)
    # .join(df_exploded["parks"])
    .rename(columns={
        "id": "park_id",
        "name": "park_name",
    })
)

df_final.head()
```
